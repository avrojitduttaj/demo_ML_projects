```python
#Price_Change_% = ((Close - Open) / Open) × 100
dly['price_change%']=((dly['Close']-dly['Open'])/dly['Open'])*100
wkly['price_change%']=((wkly['Close']-wkly['Open'])/wkly['Open'])*100
mthly['price_change%']=((mthly['Close']-mthly['Open'])/mthly['Open'])*100
```

Task 3.3 - Returns

```python
dly['Return']=dly['Close'].pct_change()*100
wkly['Return']=wkly['Close'].pct_change()*100
mthly['Return']=mthly['Close'].pct_change()*100
```

```python
print(dly.columns)
print(wkly.columns)
print(mthly.columns)
```

```python
return_stats=pd.DataFrame({'Daily':[dly['Return'].mean(),dly['Return'].median(),dly['Return'].std(),dly['Return'].min(),dly['Return'].max()],
                           'Weekly':[wkly['Return'].mean(),wkly['Return'].median(),wkly['Return'].std(),wkly['Return'].min(),wkly['Return'].max()],
                           'Monthly':[mthly['Return'].mean(),mthly['Return'].median(),mthly['Return'].std(),mthly['Return'].min(),mthly['Return'].max()]},
                          index=['Mean return','Median return','Standard deviation','Minimum return','Maximum return'])

return_stats
```

Task 3.4 - Volatility

```python
dly['Volatility']=dly['Return'].rolling(window=20).std()
wkly['Volatility']=wkly['Return'].rolling(window=4).std()
mthly['Volatility']=mthly['Return'].rolling(window=3).std()
```

```python
print(dly.columns)
print(wkly.columns)
print(mthly.columns)
```

```python
#daily
plt.figure(figsize=(12,5))
plt.plot(dly['Date'],dly['Volatility'])
plt.title('20-day rolling volatility for daily data')
plt.xlabel('Date')
plt.ylabel('Volatility')
plt.show()
```

```python
#weekly
plt.figure(figsize=(12,5))
plt.plot(wkly['Date'],wkly['Volatility'])
plt.title('4-week rolling volatility for weekly data')
plt.xlabel('Date')
plt.ylabel('Volatility')
plt.show()
```

```python
#monthly
plt.figure(figsize=(12,5))
plt.plot(mthly['Date'],mthly['Volatility'])
plt.title('3-month rolling volatility for monthly data')
plt.xlabel('Date')
plt.ylabel('Volatility')
plt.show()
```

```python
print('Highest Volatility in a day')
dly.nlargest(10,'Volatility')[['Date','Volatility']]
```

```python
print('Lowest Volatility in a week')
wkly.nsmallest(10,'Volatility')[['Date','Volatility']]
```
